```python
import numpy as np

from nsga_population import Population, ClassicalPopulation
from individual import Individual, PermutationChromosome
```

```python
n_machines = 3
n_jobs = 3
jssp_problem = np.array([
          # Technological sequence. Each row represents one job J. Each column represents the operation index.
          [[2, 3, 1],
           [1, 2, 3],
           [1, 3, 2]],
          # The second matrix contains the processing time
          [[1, 5, 2],
           [3, 3, 5],
           [2, 2, 4]]

])

test = ClassicalPopulation(N=50, decoding_method="apply_operation_based_bierwirth", n_jobs=n_jobs, n_machines=n_machines, objectives=["Makespan", "Mean Completion Time"], 
                           activate_schedule=True, tournament_size=2, mating_pool_size=50, mutation_probability=0.3, jssp_problem=jssp_problem)
```

```python
np.random.seed(0)
donator = np.repeat(np.arange(n_jobs), n_machines)[np.argsort(np.random.randint(n_jobs*n_machines, size=n_jobs*n_machines))]
print(donator)
np.random.seed(0)
receiver = np.repeat(np.arange(n_jobs), n_machines)[np.argsort(np.random.randint(n_jobs*n_machines, size=n_jobs*n_machines))]
print(receiver)
result_indiv = test.gox_crossover(donator, receiver)
print(receiver)
print(donator)
```
